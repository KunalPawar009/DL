```python
import tensorflow as tf
import matplotlib.pyplot as plt
import time

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

X_train = X_train.reshape(60000,784)
X_test = X_test.reshape(10000,784)

y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
model1 = Sequential()

model1.add(Dense(128, activation='sigmoid', input_shape=(784,)))
model1.add(Dense(64, activation='sigmoid'))
model1.add(Dense(10, activation='softmax'))

model1.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

start1 = time.time()

history1 = model1.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

end1 = time.time()

loss1, accuracy1 = model1.evaluate(X_test, y_test)

time1 = end1 - start1
model2 = Sequential()

model2.add(Dense(128, activation='tanh', input_shape=(784,)))
model2.add(Dense(64, activation='tanh'))
model2.add(Dense(10, activation='softmax'))

model2.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

start2 = time.time()

history2 = model2.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

end2 = time.time()

loss2, accuracy2 = model2.evaluate(X_test, y_test)

time2 = end2 - start2
model3 = Sequential()

model3.add(Dense(128, activation='relu', input_shape=(784,)))
model3.add(Dense(64, activation='relu'))
model3.add(Dense(10, activation='softmax'))

model3.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

start3 = time.time()

history3 = model3.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

end3 = time.time()

loss3, accuracy3 = model3.evaluate(X_test, y_test)

time3 = end3 - start3
plt.figure(figsize=(8,5))

plt.plot(history1.history['accuracy'], label='Sigmoid')
plt.plot(history2.history['accuracy'], label='Tanh')
plt.plot(history3.history['accuracy'], label='ReLU')

plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
plt.figure(figsize=(8,5))

plt.plot(history1.history['val_accuracy'], label='Sigmoid')
plt.plot(history2.history['val_accuracy'], label='Tanh')
plt.plot(history3.history['val_accuracy'], label='ReLU')

plt.title("Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
plt.figure(figsize=(8,5))

plt.plot(history1.history['loss'], label='Sigmoid')
plt.plot(history2.history['loss'], label='Tanh')
plt.plot(history3.history['loss'], label='ReLU')

plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.show()
activation = ["Sigmoid", "Tanh", "ReLU"]
training_time = [time1, time2, time3]

plt.figure(figsize=(6,4))

plt.bar(activation, training_time)

plt.title("Training Time")
plt.xlabel("Activation Function")
plt.ylabel("Seconds")

plt.show()
activation = ["Sigmoid", "Tanh", "ReLU"]
accuracy = [accuracy1, accuracy2, accuracy3]

plt.figure(figsize=(6,4))

plt.bar(activation, accuracy)

plt.title("Test Accuracy")
plt.xlabel("Activation Function")
plt.ylabel("Accuracy")

plt.show()
layers = np.arange(1,11)

sigmoid = np.exp(-0.7*layers)
tanh = np.exp(-0.4*layers)
relu = np.ones(10)

plt.figure(figsize=(8,5))

plt.plot(layers, sigmoid, marker='o', label='Sigmoid')
plt.plot(layers, tanh, marker='s', label='Tanh')
plt.plot(layers, relu, marker='^', label='ReLU')

plt.title("Gradient Flow")
plt.xlabel("Layers")
plt.ylabel("Gradient")

plt.legend()
plt.grid()

plt.show()
```
